```python
from openai import OpenAI
```

```python
client = OpenAI(base_url="http://localhost:11434/v1",api_key="ollama")
```

```python
while True:
    text = input("Enter sentence: ")
    

    if text.lower() == "exit":
        break
    language = input("Enter language (English/Hindi/French/Spanish/Malayalam etc.): ")
    response = client.chat.completions.create(model="llama3:8b",messages=[{"role": "system","content": "You are an AI grammar correction assistant."},{"role": "user","content": f"Correct the grammar of this sentence in {language}:{text}"}])
    print(response.choices[0].message.content)
```
